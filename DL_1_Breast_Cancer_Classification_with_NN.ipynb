Importing the Dependencies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

Data Collection and Preprocessing

```python
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
# loading data into data frame
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
data_frame.head()
```

```python
data_frame['label'] = breast_cancer_dataset.target
```

```python
data_frame.label.head(3)
```

```python
data_frame.shape
```

```python
data_frame.info()
```

```python
data_frame.isnull().sum()
```

```python
data_frame.describe()
```

0 --> Benign

1 --> Malignant

```python
data_frame.groupby('label').mean()
```

Seprating the features and target

```python
X = data_frame.drop('label', axis = 1)
Y = data_frame['label']
```

```python
print(X)
```

```python
print(Y)
```

```python
Y.shape
```

Splitting train-test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# standarding

scaler = StandardScaler()
X_train_std = scaler.fit_transform(X_train)
X_test_stad = scaler.transform(X_test)
```

**Building Neural Network**

```python
# importing tensorflow and keras

import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
# setting up layers of NN

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='sigmoid')
])
```

```python
# compiling the NN

model.compile(optimizer='adam',
              loss = 'sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
# traning the NN

history = model.fit(X_train_std, Y_train, validation_split=0.1, epochs=10)
```

Validation

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('E-box')

plt.legend(['traing data', 'validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('E-box')

plt.legend(['Traing data', 'Validation data'], loc = 'upper right')
```

Accuracy of the model on test data

```python
loss, accuracy = model.evaluate(X_test_stad, Y_test)
print(accuracy)
```

```python
print(X_test_stad.shape)
print(X_test_stad[0])
```

```python
Y_pred = model.predict(X_test_stad)
```

```python
print(Y_pred.shape)
print(Y_pred[0])
```

```python
# converting the prediction value into class label, 0 or 1.

Y_pred_labels = [np.argmax(i) for i in Y_pred]
print(Y_pred_labels)
```

Building the predictive system

```python
input_data = (20.57,17.77,132.9,1326,0.08474,0.07864,0.0869,0.07017,0.1812,0.05667,0.5435,0.7339,3.398,74.08,0.005225,0.01308,0.0186,0.0134,0.01389,0.003532,24.99,23.41,158.8,1956,0.1238,0.1866,0.2416,0.186,0.275,0.08902)

# Convert input_data into a NumPy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshape the NumPy array because we are predicting for one data point
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# Standardize the data using the scaler fitted on training data
input_data_std = scaler.transform(input_data_reshaped)

# Make prediction
prediction = model.predict(input_data_std)
print(prediction)

# Get the predicted class label
prediction_label = np.argmax(prediction)
print(prediction_label)

# Display the result
if prediction_label == 0:
    print('The tumor is Benign')
else:
    print('The tumor is Malignant')
```
